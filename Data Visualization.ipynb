```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('c:/Users/Lenovo/Downloads/archive (4)/healthcare-dataset-stroke-data.csv')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (8, 5)   
```

```python
sns.countplot(x='stroke', data=df, palette='Set2')
plt.title("How many people had a stroke?")
plt.xlabel("Stroke (0 = No, 1 = Yes)")
plt.ylabel("Number of people")

# showing numbers on bars
for p in plt.gca().patches:
    plt.gca().annotate(f'{p.get_height()}', 
                       (p.get_x() + p.get_width()/2., p.get_height()), 
                       ha='center', va='bottom', fontsize=11)

plt.show()
```

```python
#age and stroke
sns.boxplot(x='stroke', y='age', data=df, palette='Set2')
plt.title("Age of people with and without stroke")
plt.xlabel("Stroke (0 = No, 1 = Yes)")
plt.ylabel("Age (years)")
plt.show()
```

```python
#glocose level and stroke
sns.boxplot(x='stroke', y='avg_glucose_level', data=df, palette='Set2')
plt.title("Average Glucose Level vs Stroke")
plt.xlabel("Stroke (0 = No, 1 = Yes)")
plt.ylabel("Glucose level (mg/dL)")
plt.show()
```

```python
#high blood pressure & heart disease
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(x='hypertension', hue='stroke', data=df, ax=axes[0], palette='Set2')
axes[0].set_title("Stroke by Hypertension")

sns.countplot(x='heart_disease', hue='stroke', data=df, ax=axes[1], palette='Set2')
axes[1].set_title("Stroke by Heart Disease")

plt.tight_layout()
plt.show()
```

```python
#smoking status
sns.countplot(x='smoking_status', hue='stroke', data=df, palette='Set2')
plt.title("Stroke by Smoking Status")
plt.xlabel("Smoking")
plt.ylabel("Number of people")
plt.xticks(rotation=20)
plt.show()
```
